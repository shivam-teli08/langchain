So first of all we import model class using specific provider or init_chat_models

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

initlialize model here 

```python
model = init_chat_model(model="codestral-2508", model_provider="mistralai",temperature=0.3)
```

invoke and print response 

```python
result = model.invoke("Give me detail info about Mistral AI and its products.")
print(result)
```

so we use stream to see running outputs means it helps us to see output as it is generated 

```python
for result in model.stream("Give me detail info about Mistral AI and its products."):
    print(result.content, end="")
```

chat models


```python
from langchain_mistralai import ChatMistralAI
from dotenv import load_dotenv
import os 
load_dotenv()  # Load environment variables from .env file
if(key := os.getenv("MISTRAL_API_KEY")) is None:
    raise ValueError("MISTRALAI_API_KEY environment variable is not set.")
```

initlialize model

```python
chatModelInitialization = ChatMistralAI(model="codestral-2508", temperature=0.3)
```

invokation

```python
messages=[
    ("system","You are a helpful assistant that translates English to French."),
    ("user","Translate this sentence from English to French. I love programming."),
]
try:
    chatModelResponse = chatModelInitialization.invoke(messages)
    print(chatModelResponse.content)
except Exception as e:
    print(f"An error occurred: {e}")
```
