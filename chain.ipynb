Chain describes workflow that happens one after another 

```python
#type 1 : sequential chain 
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_mistralai import ChatMistralAI
from langchain_core.output_parsers import StrOutputParser
load_dotenv()
```

```python
prompt = PromptTemplate.from_template(
    "Explain the how this topic works {topic}"
)
llm = ChatMistralAI(model_name="codestral-2508")
parser = StrOutputParser()
```

```python
chain = prompt | llm | parser 
result = chain.invoke({"topic": "LangChain"})
print(result)
```

Parallel chain 

```python
from langchain_core.runnables import RunnableParallel 


prompt1 = PromptTemplate.from_template(
    "Explain the how this topic works {topic1}"
)
prompt2 = PromptTemplate.from_template(
    "Explain the how this topic works {topic2}"
)
parallel_chain = RunnableParallel(
    topic1 = prompt1 | llm | parser,
    topic2 = prompt2 | llm | parser
)
result = parallel_chain.invoke({"topic1": "LangChain", "topic2": "LangChain"})
print(result)
```

Conditional chain 

```python
from langchain_core.runnables import RunnableBranch
 
complanit_prompt = PromptTemplate.from_template(
    "Explain the how this topic works {topic} but i have a complaint wbout it ?"
)
general_prompt = PromptTemplate.from_template(
    "Explain the how this topic works {topic} ?"
)
question_prompt = PromptTemplate.from_template(
    "Explain the how this topic works but first ask a question about {topic} ?"
)
```

```python
complaint_chain = complanit_prompt | llm | parser
general_chain = general_prompt | llm | parser
question_chain = question_prompt | llm | parser
```

```python
branch_chain = RunnableBranch(
    (lambda x: "complaint" in x["topic"].lower(),complaint_chain),
    (lambda x: "question" in x["topic"].lower(),question_chain),
    general_chain
)

result = branch_chain.invoke({"topic": "LangChain"})
print(result)
```
