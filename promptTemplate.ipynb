A PromptTemplate is a reusable text blueprint for the prompt you send to an LLM. You write the prompt once with placeholders (variables) in curly braces, and at runtime you fill in the actual values.

```python
from dotenv import load_dotenv 
from langchain_core.prompts import PromptTemplate

load_dotenv()
```

Template intialization

```python
template = PromptTemplate(
    input_variables=["product"],
    template="What is a good name for a company that makes {product}?"
)
# template2 = PromptTemplate.from_template("What is a good name for a company that makes {product}?")
```

Run template 

```python
prompt_init = template.format(product="colorful socks")
# prompt_init2 = template.invoke({"product": "colorful socks"})
from langchain.chat_models import init_chat_model

llm = init_chat_model(model="codestral-2508", model_provider="mistralai",temperature=0.3)
for message in llm.stream(prompt_init):
    print(message.content, end="")
```

ChatPromptTemplate 

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages(
    [
        ("system","You are a helpful assistant that translates {input_language} to {output_language}."),
        ("human","Translate the following text: {text}"),
    ]
)
```

fill values 

```python
chat_prompt_init = chat_template.format_messages(
    input_language="English",
    output_language="French",
    text="I love programming."
)
for message in llm.stream(chat_prompt_init):
    print(message.content, end="")
    
```
